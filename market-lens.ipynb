```python
!pip install yfinance -q
import yfinance as yf
import pandas as pd
import numpy as np

df = yf.download("AAPL", start="2023-01-01", end="2024-01-01")
df.head()
```

```python
df['Return'] = df['Close'].pct_change()
df['MA20'] = df['Close'].rolling(20).mean()
df[['Close', 'Return', 'MA20']].tail()
```

```python
print("Average daily return:", df['Return'].mean())
print( "Best day:", df['Return'].max())
print("Worst day:" df['Return']).min())
df[['Close', 'MA20']].plot(figsize=(12,5), title="AAPL - price vs 20-day average")
```
